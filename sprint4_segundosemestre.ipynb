```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score

df = pd.read_csv("/content/Renewable_Energy_Data.csv")

le = LabelEncoder()
df['Energy_Class_Encoded'] = le.fit_transform(df['Energy_Class'])

df_numeric = df.select_dtypes(include=['number'])
matriz_corr = df_numeric.corr()

print("--- MATRIZ DE CORRELAÇÃO (Em relação ao Target) ---")
print(matriz_corr['Energy_Class_Encoded'].sort_values(ascending=False))

"""
01) d) Resposta:
Analisando a matriz de correlação, optamos por descartar as features 'Wind_Speed_m_s',
'Solar_Radiation_kWh_m2' e 'Combined_Weather_Index', pois possuem as correlações mais
próximas de zero, gerando apenas ruído. Mantivemos as outras, com destaque para
'Efficiency_Ratio', que possui a correlação mais forte com o target.
"""

X = df_numeric.drop(columns=[
    'Energy_Class_Encoded',
    'Wind_Speed_m_s',
    'Solar_Radiation_kWh_m2',
    'Combined_Weather_Index'
])
y = df['Energy_Class_Encoded']

"""
02) a) e b) Resposta:
O modelo escolhido é a Regressão Logística (LogisticRegression do scikit-learn).
Ele é linear porque sua fronteira de decisão (boundary condition) é definida por um hiperplano
(uma reta matemática em múltiplas dimensões) que separa as classes no espaço de features
através de uma combinação linear dos pesos e variáveis.
"""

X_train_40, X_test_40, y_train_40, y_test_40 = train_test_split(X, y, test_size=0.40, random_state=42)
X_train_15, X_test_15, y_train_15, y_test_15 = train_test_split(X, y, test_size=0.15, random_state=42)

modelo_40 = LogisticRegression(max_iter=5000)
modelo_40.fit(X_train_40, y_train_40)

modelo_15 = LogisticRegression(max_iter=5000)
modelo_15.fit(X_train_15, y_train_15)

y_pred_40 = modelo_40.predict(X_test_40)
y_pred_15 = modelo_15.predict(X_test_15)

print("\nCENÁRIO 1 (40% de Teste)")
print("Matriz de Confusão:\n", confusion_matrix(y_test_40, y_pred_40))
print(f"Acurácia: {accuracy_score(y_test_40, y_pred_40):.4f}")
print(f"Precisão: {precision_score(y_test_40, y_pred_40, average='weighted', zero_division=0):.4f}")
print(f"Recall:   {recall_score(y_test_40, y_pred_40, average='weighted', zero_division=0):.4f}")

print("\nCENÁRIO 2 (15% de Teste)")
print("Matriz de Confusão:\n", confusion_matrix(y_test_15, y_pred_15))
print(f"Acurácia: {accuracy_score(y_test_15, y_pred_15):.4f}")
print(f"Precisão: {precision_score(y_test_15, y_pred_15, average='weighted', zero_division=0):.4f}")
print(f"Recall:   {recall_score(y_test_15, y_pred_15, average='weighted', zero_division=0):.4f}")

print("""
03) a) d) e e) Respostas:
Comentário dos resultados: A performance foi boa e bastante realista (atingindo médias de 75% a 80% de acurácia).
Considerando que as features originais possuíam correlações muito baixas com o target (máximo de -0.16), o modelo
conseguiu extrair padrões muito satisfatórios. Analisando as matrizes de confusão, nota-se um comportamento interessante:
o modelo é excelente em distinguir a Classe 0 da Classe 1 (nenhum falso positivo entre elas). Os erros se concentram
na Classe 2, que apresenta sobreposição matemática e acaba sendo confundida com as outras duas. Ainda assim,
a maioria das predições está na diagonal principal.

Comparação dos cenários: Um test set menor (15%) aumenta a quantidade de dados para o modelo
aprender no treino (85%), o que tende a melhorar o aprendizado em datasets menores, mas reduz a
garantia de que ele funciona bem no mundo real (desvantagem). Já o test set maior (40%) garante
uma validação mais severa sobre o poder de generalização do modelo, porém retira dados essenciais
do treinamento, o que pode causar underfitting. No nosso caso, o cenário com 40% curiosamente performou
um pouco melhor (80% vs 75%), indicando que as proporções das classes no sorteio dessa semente beneficiaram a validação.
""")
```
